```python
def show_wgraph(G):
    plt.figure()    
    pos = nx.spring_layout(G)
    weight_labels = nx.get_edge_attributes(G,'weight')
    nx.draw(G,pos,font_color = 'white', font_size=6, node_shape = 's', node_size=70, with_labels = True,)
    output = nx.draw_networkx_edge_labels(G,pos,edge_labels=weight_labels,font_size=6)
```
